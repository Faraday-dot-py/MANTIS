```python
def extract_filename(url):
    # Remove any leading or trailing whitespace
    url = url.strip()
    
    # Extract the file name using os.path.basename
    file_name = os.path.basename(url)
    
    return file_name
```

```python
os.makedirs('ASL_Videos', exist_ok=True)

# Function to download a video
def download_video(url):
    response = requests.get(url, stream=True)
    fileName = extract_filename(url)
    if response.status_code == 200:
        file_path = os.path.join('ASL_Videos', fileName)
        if not "definition" in fileName:
            with open(file_path, 'wb') as f:
                for chunk in response.iter_content(1024):
                    f.write(chunk)
            print(f"Downloaded: {fileName}")
    else:
        print(f"Failed to download: {fileName}")

# Download all videos
for video in videoUrls:
    download_video(video)
```

```python
import cv2
import mediapipe as mp

# Initialize Mediapipe Hand Tracking and Drawing Utilities
mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# Input and Output paths
input_video_path = r'C:\Users\awebb\Documents\Programming\Python\MANTIS\training_data\WLASL\wlasl_videos\a_lot\sample_0.mp4'
output_video_path = r'C:\Users\awebb\Documents\Programming\Python\MANTIS\training_data\WLASL\wlasl_tmaps\sample_0.mp4'

# Open the input video
cap = cv2.VideoCapture(input_video_path)

# Get video properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Define the codec and create VideoWriter object to save the output
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # You can use other codecs like 'XVID'
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# Initialize Mediapipe's Hand object
hands = mp_hands.Hands(static_image_mode=False, max_num_hands=2, min_detection_confidence=0.5)

# Process video frame by frame
while cap.isOpened():
    ret, frame = cap.read()
    
    if not ret:
        break  # End of video
    
    # Convert the BGR image to RGB (as required by Mediapipe)
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    # Process the frame to detect hand landmarks
    results = hands.process(frame_rgb)
    
    # If landmarks are detected, draw them on the frame
    if results.multi_hand_landmarks:
        for hand_landmarks in results.multi_hand_landmarks:
            mp_drawing.draw_landmarks(
                frame,  # Draw on the original BGR frame
                hand_landmarks,
                mp_hands.HAND_CONNECTIONS,
                mp_drawing.DrawingSpec(color=(0, 255, 0), thickness=2, circle_radius=2),  # Style for landmarks
                mp_drawing.DrawingSpec(color=(0, 0, 255), thickness=2)  # Style for connections
            )
    
    # Write the frame with landmarks drawn to the output video
    out.write(frame)
    
    # Optionally, display the frame in real-time (if you want to see the output while processing)
    cv2.imshow('Hand Tracking', frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break  # Exit if 'q' is pressed

# Release resources
cap.release()
out.release()
hands.close()
cv2.destroyAllWindows()
```
