```python
from tqdm import tqdm
import os
import shutil
```

```python
srcDir = r'C:/Users/awebb/Downloads/lsa64_cut/all_cut/'
outDir = r'C:/Users/awebb/Documents/Programming/Python/MANTIS/training_data/lsa64/'
```

```python
# print(words)
# print(wordVideos)

with tqdm(total=len(words)) as pbar:
    for word, wordVideoList in zip(words, wordVideos):
        # print(f"Doing {word}")
        savedir = outDir+word.lower().strip().replace(' ', '_')
        if not os.path.exists(savedir):
            os.mkdir(savedir)

        # # print(wordVideo)
            for file in wordVideoList:
                shutil.copyfile(srcDir+file, savedir+r'/sample_'+str(len(os.listdir(savedir)))+'.mp4')
        pbar.update(1)
```
