```python
from TemporalMap import TemporalMap
import os
import time
import numpy as np
```

```python
tmap = TemporalMap()
```

```python
tmap.loadHandLandmarker()

tmapArray = []
times = []

# print(os.listdir('./captured_images/l2r'))
for images in os.listdir('./captured_images/l2r'):
    print(f'Creating tmap for {images}')
    stime = time.time()
    try:
        tmapArray.append(tmap.createTemporalMapFromDir(f'./captured_images/l2r/{images}')[0])
        times.append(time.time()-stime)
        print(f'Finished tmap for {images}')
    except IndexError:
        print(f"Couldn't get the tmap for {images}")

print(f'Average time: {np.average(times)}')
# for cap in os.listdir('./captured_images'):

# tmap.createTemporalMapFromDir('./captured_images')

# tmap.displayResults(os.listdir('./output_images'))
```
