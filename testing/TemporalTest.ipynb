```python
from TemporalMap import TemporalMap

tmap = TemporalMap(CONTEXT_WINDOW=30, DISPLAY_CAMERA_VIEW=True, RENDER_TMAPS=True)

tmap.setup()
```

```python
import numpy as np
from tensorflow.keras.preprocessing import image
from IPython.display import clear_output
```

```python
def preprocess_image(img, img_size=(30, 42)):
    # Load and resize the image
    img_array = image.img_to_array(img)
    img_array = img_array / 255.0  # Normalize pixel values if necessary
    
    # Add a batch dimension
    img_array = np.expand_dims(img_array, axis=0)  # Shape becomes (1, 30, 42, 3)
    return img_array
```

```python
import mediapipe as mp

loadedTmap = mp.Image.create_from_file(f'output_images/r2l/sample_14.jpg')

preprocessedHandMap = preprocess_image(tmap.tmap)

print(model.predict(preprocessedHandMap))


```
