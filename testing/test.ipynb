```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np

def display_image_cropped(file_path):
    try:
        # Read the image file
        img = mpimg.imread(file_path)
        
        # Determine the midpoint of the width
        midpoint = img.shape[1] // 2
        
        # Crop the image by slicing to keep only the right half
        img_cropped = img[:, midpoint:]

        # Display the cropped image
        plt.imshow(img_cropped)
        plt.xlabel('Hand Index')  # Add X-axis label
        plt.ylabel('Age')  # Add Y-axis label
        plt.show()
        
    except FileNotFoundError:
        print(f"Error: The file at '{file_path}' was not found.")
    except IOError:
        print(f"Error: Could not read the file at '{file_path}'.")


if __name__ == "__main__":
    # Replace 'your_image_path_here.jpg' with your actual image file path
    file_path = r'C:\Users\awebb\Documents\Programming\Python\MANTIS\output_images\l2r\sample_0.jpg'
    display_image_cropped(file_path)
```
